```python
import matplotlib.pyplot as plt
```

### Bar plot
1. Compairing category
2. showing frequencies or count

```python
products=["Hot Coffee","Cold Coffee","Tea","Salad"]
summer=[30,90,80,50]
```

```python
plt.figure(figsize=(7,3)) #width,height
plt.bar(products,summer)
plt.title("Sales of products for summer")
plt.xlabel("product")
plt.ylabel("count")
plt.show()
          
```
