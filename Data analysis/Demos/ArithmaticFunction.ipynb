### Mathematical Arithmatic Function

```python
import numpy as np
```

```python
arr1=np.arange(1,10).reshape((3,3))
arr2=np.arange(3,12).reshape((3,3))
arr1, arr2
```

```python
np.add(arr2,arr1)
```

```python
arr2+arr1
```

```python
arr3=np.array([10,20,30])
arr3
```

```python
np.subtract(arr2,arr1)
```

```python
arr1-arr2
```

```python
arr1+arr3
```

```python
np.dot(arr1,arr2)
```

```python
np.sqrt(arr1)
```

```python
np.log(arr1)
```

```python
np.log10(arr1)
```

```python
np.multiply(arr1,arr2)
```

```python
np.divide(arr1,arr2)
```

```python
arr1/arr2
```

```python
arr1+arr2
```

```python
np.concatenate([arr1,arr2]) #No of columns should be same
```

```python
np.concatenate([arr1,arr2],axis=1) #No of rows should be same
```
