## Merge

```python
emp=pd.DataFrame({"id":[101,102,103,104],"Name":["abc","xyz","pqr","mno"],"did":[1,2,1,4]})
emp
```

```python
dept=pd.DataFrame({"did":[1,2,3],"dname":["Admin","IT","Sales"]})
dept
```

```python
pd.merge(emp,dept,how="inner")
```

```python
pd.merge(emp,dept,how="left")
```

```python
pd.merge(emp,dept,how="right")
```

```python
pd.merge(emp,dept,left_on="did",right_on="dept_id")
```
