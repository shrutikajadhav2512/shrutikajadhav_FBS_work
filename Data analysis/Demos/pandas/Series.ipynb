### Series
1. pd.Series()
2. Heterogeneous
3. Ordered
4. Mutable,Size-Immutable
5. Duplication allowed
6. Customized indexing available

```python
s1=pd.Series([10,20,30,"abc"])
s1
```

```python
s1=pd.Series([10,20,30,40,50,60],index=['a','b','c','d','e','f'])
s1
```

```python
s2=pd.Series([590,460,900,20,150,25],index=['a','c','b','d','f','e'])
s2
```

```python
s1+s2
```
