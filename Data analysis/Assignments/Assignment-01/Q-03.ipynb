#### Create a 1d-array of mobile phones prices

```python
import numpy as np
```

```python
arr=np.array([6999,7500,11999,27899,14999,9999])
arr
```

#### Convert this array to float type

```python
arr=arr.astype(float)
arr
```

#### Append a new mobile having price of 13999 Rs. to this array

```python
arr=np.append(arr,13999)
arr
```

### Reverse this array of mobile phones prices

```python
arr1=arr[::-1]
arr1
```

### Apply GST of 18% on mobile phones prices and update this array

```python
arr3=arr*1.18
arr3
```

### Sort the array in descending order of price

```python
arr=np.sort(arr)[::-1]
arr
```

### What is the average mobile phone price

```python
avg=np.average(arr)
avg
```

### What is the difference b/w maximum and minimum price

```python
max_price=np.max(arr)
max_price
```

```python
min_price=np.min(arr)
min_price
```

```python
diff=max_price-min_price
diff
```
