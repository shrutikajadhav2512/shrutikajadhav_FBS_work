Refer to the tips.csv dataset. The dataset contains total 244 tips(data about customers
served by a restaurant) across 7 different columns.
1. totbill - Total bill (cost of the meal), including tax, in US dollars
2. tip - Tip (gratuity) in US dollars
3. sex - Sex of person paying for the meal (0=male, 1=female)
4. smoker - Smoker in party? (0=No, 1=Yes)
5. day - Sun, Sat, Thu, Fri
6. time - Night and Day
7. size - Size of the party

### Plot a day wise boxplot of totbill column.

```python
print(df.columns)
```

```python
sns.boxplot(x="day", y="total_bill", data=df)

plt.title("Day-wise Boxplot of Total Bill")
plt.xlabel("day")
plt.ylabel("total_bill")
plt.show()
```

### Draw a scatter plot of tip and totalbill and also allot the points according to the sex

```python
sns.scatterplot(data=df,x="total_bill",y="tip",hue="sex")
plt.title("Scatter Plot of Total Bill vs Tip by Sex")
plt.xlabel("total_bill")
plt.ylabel("tip")
plt.show()
```

### Plot a hitsogram of totbill for 50 bins

```python
sns.histplot(data=df, x="total_bill", bins=50)

plt.title("Histogram of Total Bill")
plt.xlabel("total_bill")
plt.ylabel("Frequency")
plt.show()
```

### Plot a heatmap to represent entire dataset's correlation

```python
corr=df.corr(numeric_only=True)
plt.figure(figsize=(7,3))
plt.imshow(corr)
plt.title('Correlation Heatmap')
plt.colorbar()
plt.show()
```
