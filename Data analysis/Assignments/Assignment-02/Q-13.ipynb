### 13. Create a 4x4 array of random integers. Find the maximum, minimum, and mean of the entire array.

```python
import numpy as np
```

```python
arr=np.random.randint(1,50,(4,4))
arr
```

```python
maximum=np.max(arr)
maximum
```

```python
minimum=np.min(arr)
minimum
```

```python
mean_arr=np.mean(arr)
mean_arr
```
