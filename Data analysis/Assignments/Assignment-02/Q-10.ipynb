### Create a 2x2 matrix A and a 2x2 matrix B. Perform matrix multiplication using np.dot() or the @ operator.

```python
import numpy as np
```

```python
A=np.arange(1,5).reshape((2,2))
A
```

```python
B=np.arange(5,9).reshape((2,2))
B
```

#### np.dot()

```python
np.dot(A,B)
```

#### @

```python
A@B
```
