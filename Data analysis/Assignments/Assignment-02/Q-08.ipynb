### Create two arrays of shape (2, 3): arr1 and arr2. Perform element-wise addition,subtraction, multiplication, and division on them.

```python
import numpy as np
```

```python
arr1=np.arange(1,7).reshape((2,3))
arr1
```

```python
arr2=np.arange(10,16).reshape((2,3))
arr2
```

#### Adition

```python
np.add(arr1,arr2)
```

```python
arr1+arr2
```

#### Substraction

```python
np.subtract(arr1,arr2)
```

```python
np.subtract(arr2,arr1)
```

```python
arr1-arr2
```

```python
arr2-arr1
```

#### Multiplication

```python
np.multiply(arr1,arr2)
```

```python
arr1*arr2
```

#### Division

```python
np.divide(arr1,arr2)
```

```python
np.divide(arr2,arr1)
```

```python
arr1/arr2
```

```python
arr2/arr1
```
