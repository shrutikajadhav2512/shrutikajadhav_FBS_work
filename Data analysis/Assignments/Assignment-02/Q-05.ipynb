### Create a 3x3 array of integers and access:
1. The second element of the first row.
2. The third column of the 2nd row.
3. A subarray with the first two rows and last two columns.

```python
import numpy as np
```

```python
arr=np.arange(1,10).reshape((3,3))
arr
```

#### The second element of the first row

```python
arr[0][1]
```

#### The third column of the 2nd row

```python
arr[1][2]
```

#### A subarray with the first two rows and last two columns.

```python
arr[0:2,1:3]
```
