## 1. Create a DataFrame:
o Load the above dataset into a Pandas DataFrame.

```python
data = {"Order ID": [1001,1002,1003,1004,1005,1006,1007,1008],
    "Customer": ["Alice","Bob","Charlie","Alice","David","Charlie","Bob","Alice"],
    "Product": ["Laptop","Phone","Chair","Table","Laptop","Phone","Chair","Phone"],
    "Category": ["Electronics","Electronics","Furniture","Furniture","Electronics","Electronics","Furniture","Electronics"],
    "Quantity": [2,1,4,1,1,2,2,1],
    "Price": [800,500,100,250,800,500,100,500],
    "Total Sales": [1600,500,400,250,800,1000,200,500]}
df = pd.DataFrame(data)
df
```

### 2. Add a New Column:
o Add a column Discount that gives a 10% discount on total
sales.

```python
df["Discount"] = df["Total Sales"] * 0.10
df
```

```python
new_row = {"Order ID": 1009,
    "Customer": "Eve",
    "Product": "Tablet",
    "Category": "Electronics",
    "Quantity": 1,
    "Price": 300,
    "Total Sales": 300}
df.loc[len(df)] = new_row
df
```

```python
electronics_sales = df[df["Category"] == "Electronics"]
electronics_sales
```

### 5. Selection:
o Select the Customer and Total Sales columns.

```python
select=df[['Customer','Total Sales']]
select
```

### 6. Grouping:
o Group the data by Customer and calculate the total sales per
customer.

```python
group_sale=df.groupby('Customer')['Total Sales'].sum()
group_sale
```

### 7. Sorting:
o Sort the DataFrame based on Total Sales in descending
order.

```python
sorted_df = df.sort_values(by="Total Sales", ascending=False)
sorted_df
```

### 8. Apply Function:
o Apply a lambda function to create a new column Sales
Category, where values are labeled as "High" if Total Sales >
1000, otherwise "Low".

```python
df['Sales Category']= df['Total Sales'].apply(lambda x:'High'if x>1000 else'Low')
```
