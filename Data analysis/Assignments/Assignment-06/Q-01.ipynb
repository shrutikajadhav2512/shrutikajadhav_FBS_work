## Dataset: The Cars93 dataset contains information about different car models, including their manufacturer, price, horsepower, MPG, and other attributes. You will use this dataset for the following tasks.

### Bar Chart:
1. Create a bar chart to compare the average price of cars by manufacturer.
2. Label the axes and add a suitable title.
3. Use different colors for different manufacturers.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data=pd.read_csv("C:/Users/DELL/Desktop/FBS/Data analysis/Demos/Datasets-20260831T171242Z-1-001/Datasets/cars93.csv")
data.head()
```

```python
data.describe()
```

```python
data.columns
```

```python
avg_price=data['Price'].mean()
avg_price
```

```python
avg_price = data.groupby('Manufacturer')['Price'].mean()
colors = plt.cm.tab20(range(len(avg_price)))
plt.bar(avg_price.index, avg_price.values, color=colors)
plt.title('Average Price of Cars by Manufacturer')
plt.ylabel('Average Price')
plt.xlabel('Manufacturer')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### Line Chart:
1. Create a line chart showing the trend of horsepower for different types of cars.
2. Include proper legends, labels, and a title

```python
avg_hp = data.groupby('Type')['Horsepower'].mean()
plt.plot(avg_hp.index, avg_hp.values, marker='o',
         color='blue', label='Average Horsepower')
plt.title('Horsepower Trend for Different Types of Cars')
plt.xlabel('Car Type')
plt.ylabel('Average Horsepower')
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

### Pie Chart:
1. Create a pie chart showing the proportion of different car types in the dataset.
2. Include percentages in the chart.

```python
car_counts = data['Type'].value_counts()
plt.pie(car_counts,labels=car_counts.index,autopct='%1.1f%%',startangle=90,colors=plt.cm.Set3.colors)
plt.title('Proportion of Different Car Types')
plt.axis('equal')
plt.show()
```

```python
plt.scatter(data["Horsepower"], data["MPG.city"], color="blue", marker="o")
plt.xlabel("Horsepower")
plt.ylabel("MPG (Miles Per Gallon)")
plt.title("Relationship Between Horsepower and MPG")
plt.show()
```

### Line Chart:
1. Create a line chart showing the trend of horsepower for different types of cars.
2. Include proper legends, labels, and a title

```python
horsepower_trend = data.groupby('Type')['Horsepower'].mean()
plt.figure(figsize=(8, 5))
plt.plot(
    horsepower_trend.index,
    horsepower_trend.values,
    marker='o',
    linestyle='-',
    color='blue',
    label='Average Horsepower'
)
plt.xlabel('Type of Car')
plt.ylabel('Average Horsepower')
plt.title('Horsepower Trend for Different Types of Cars')
plt.show()
```
